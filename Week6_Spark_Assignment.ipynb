# Q1. Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application.

## Answer

**Driver:**  
The Driver is the main program of a Spark application. It creates the SparkSession, schedules tasks, and collects results from executors.

**Cluster Manager:**  
The Cluster Manager manages cluster resources and allocates executors to Spark applications.

**Executor:**  
Executors run tasks assigned by the Driver, process data, cache intermediate results, and send results back to the Driver.

# Q2. How does Spark’s Lazy Evaluation strategy improve performance when chain-processing large datasets?

## Answer

Spark does not execute transformations immediately. Instead, it records all transformations as a Directed Acyclic Graph (DAG). Execution starts only when an action such as `show()` or `collect()` is called. This allows Spark to optimize the execution plan, reduce unnecessary computations, and improve overall performance.

# Q4. What is the difference between CSV and Parquet in terms of storage and why does it matter for performance?

## Answer

| CSV | Parquet |
|------|----------|
| Row-based storage | Columnar storage |
| Larger file size | Compressed file size |
| Slower performance | Faster performance |
| No schema support | Stores schema |

Parquet improves performance because Spark reads only the required columns, reducing disk I/O and memory usage.

# Q5. Given a DataFrame df, write a query to select the columns product_id and price where the category is 'Electronics'.

```python
from pyspark.sql.functions import col

df.filter(col("category") == "Electronics") \
  .select("product_id", "price")
```

# Q7. How does Spark use the Lineage Graph (DAG) to provide fault tolerance if a worker node fails?

## Answer

Spark stores all transformations as a Lineage Graph (DAG). If a worker node fails, Spark recomputes only the lost partitions using the DAG instead of processing the entire dataset again. This provides efficient fault tolerance.

# Q8. Write a query to filter a DataFrame df_orders for rows where the status is 'Completed' AND the amount is greater than 1000.

```python
from pyspark.sql.functions import col

df_orders.filter(
    (col("status") == "Completed") &
    (col("amount") > 1000)
)
```

# Q9. Explain the concept of Predicate Pushdown in Parquet.

## Answer

Predicate Pushdown allows Spark to send filter conditions directly to the Parquet file reader. Only the required rows are loaded into memory, reducing disk I/O, memory usage, and improving query performance.

# Q11. What is the difference between Transformations and Actions? Provide two examples of each.

## Answer

### Transformations

Transformations create a new DataFrame or RDD and are evaluated lazily.

Examples:
- filter()
- select()

### Actions

Actions trigger the execution of transformations and return results.

Examples:
- show()
- collect()

# Q12. Load a Parquet file, remove rows where user_id is null, and save the result as a CSV.

```python
from pyspark.sql.functions import col

df = spark.read.parquet("path/to/input")

df.filter(col("user_id").isNotNull()) \
  .write \
  .mode("overwrite") \
  .option("header", True) \
  .csv("path/to/output")
```

# Q13. What is the difference between Client Mode and Cluster Mode?

## Answer

| Client Mode | Cluster Mode |
|--------------|--------------|
| Driver runs on the client machine | Driver runs inside the cluster |
| Suitable for development and testing | Suitable for production |
| Client must remain connected | Client can disconnect after job submission |
| Easier debugging | Better scalability and reliability |

# Q15. Why is it safer to use .show(5) instead of .collect() on a multi-terabyte dataset?

## Answer

`.show(5)` displays only the first five rows of the dataset, making it safe for exploring large datasets.

`.collect()` transfers the entire dataset to the Driver's memory. For multi-terabyte datasets, this can cause excessive memory usage, application slowdown, or even application failure due to insufficient memory.
